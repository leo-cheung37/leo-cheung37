```python
import pandas as pd
df = pd.read_csv('bakery_sales_dataset.csv',encoding="utf-8")
df.head()
```

## Dataset description

```python
#Basic information
df.info()
```

```python
#Change Date to date format
df['Date']=pd.to_datetime(df['Date'],errors='coerce')
```

```python
#Null count
df.isnull().sum().loc[lambda x:x>0]
```

```python
#Get column information of Categorical & Numeric
df.dtypes #we can see object, int64, float64
numeric_cols=df.select_dtypes(include='number').columns.tolist()
object_cols=df.select_dtypes(exclude='number').columns.tolist()
```

```python
#Numeric
for col_name in numeric_cols:
    curr_col=df[col_name]
    print(f'<---{col_name}---->')
    print(f"Unique count: {curr_col.nunique()} - Range: {curr_col.describe()['min']} - {curr_col.describe()['max']}")
    print(f"Mean: {curr_col.describe()['mean']:.2f} / Mode: {curr_col.mode()[0]:.2f} / Median: {curr_col.median():.2f} / Standard Deviation: {curr_col.describe()['std']:.2f}")
    print(f"1st Quartile: {curr_col.quantile(0.25):.2f} - 2nd Quartile: {curr_col.quantile(0.5):.2f} - 3th Quartile: {curr_col.quantile(0.75):.2f}")
```

```python
#Categorical
for col_name in object_cols:
    curr_col=df[col_name]
    print(f'<---{col_name}---->')
    print(f'dtype:{curr_col.dtype} - Null count: {curr_col.isna().sum()} ({(curr_col.isna().sum()/curr_col.size)*100:.2f}%) - Unique count: {curr_col.nunique()}')
    unique_list=sorted(curr_col.dropna().unique().tolist())
    print(f"Top 10 Values: {','.join(map(str,unique_list[:10]))}")
    
```
